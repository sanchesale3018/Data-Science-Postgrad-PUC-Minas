```python
import pandas as pd
```

```python
dataset=pd.read_excel("dataset.xlsx")
dataset
```

```python
from transformers import pipeline as pip

analisador = pip("sentiment-analysis", model = "nlptown/bert-base-multilingual-uncased-sentiment", framework = "pt")
resultados = analisador(dataset["Texto"].tolist())
notas = []
for resultado in resultados:
    notas.append(int(resultado['label'].split()[0]))
    
dataset["Nota"] = notas
```
